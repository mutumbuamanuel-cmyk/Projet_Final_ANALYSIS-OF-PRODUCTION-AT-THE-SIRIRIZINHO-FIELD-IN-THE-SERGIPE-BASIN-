```python
import pandas as pd
```

```python
anos = [2026, 2025, 2024, 2023]
meses = []
for i in range(1,13):
  meses.append(i)
arquivos=[]
formato = '.csv'

for ano in anos:
  for mes in meses:
    if ano == 2026 and mes >= 1 and mes < 8:
      data = str(ano) + "_0" + str(mes)
      arquivos.append(data)
    elif ano == 2025 and mes >= 1 and mes < 10:
      data = str(ano) + "_0" + str(mes)
      arquivos.append(data)
    elif ano == 2025 and mes >= 10:
      data = str(ano) + "_" + str(mes)
      arquivos.append(data)
    elif ano == 2024 and mes >= 1 and mes < 10:
      data = str(ano) + "_0" + str(mes)
      arquivos.append(data)
    elif ano == 2024 and mes >= 10:
      data = str(ano) + "_" + str(mes)
      arquivos.append(data)
    elif ano == 2023 and mes >= 1 and mes < 10:
      data = str(ano) + "_0" + str(mes)
      arquivos.append(data)
    elif ano == 2023 and mes >= 10:
      data = str(ano) + "_" + str(mes)
      arquivos.append(data)

# DataFrame base que vai acumular todos os poços
df_final = None

for arquivo in arquivos:
    nome_arquivo_entrada = arquivo + '_producao_Terra' + formato

    try:
        # Lê o arquivo do mês atual
        df = pd.read_csv(nome_arquivo_entrada, sep=";", encoding='latin-1')
        df = df.drop(columns=['Número do Contrato'])
        df = df[df['Campo'] == "SIRIRIZINHO"]

        # Seleciona apenas as colunas de interesse
        df_mes = df[['Nome Poço ANP', 'Petróleo (bbl/dia)', 'Gás Natural (Mm³/dia) Total', 'Água (bbl/dia)']].copy()

        # Renomeia as colunas especificando o mês/arquivo correspondente
        df_mes.rename(columns={
            'Petróleo (bbl/dia)': 'O_bpd_' + arquivo,
            'Gás Natural (Mm³/dia) Total': 'GN_Mcm_' + arquivo,
            'Água (bbl/dia)': 'A_bpd_' + arquivo
        }, inplace=True)

        # Se for o primeiro arquivo, ele vira a base
        if df_final is None:
            df_final = df_mes
        else:
            # Se já temos uma base, fazemos um 'merge' (junção) lateral baseado no Poço ANP
            df_final = pd.merge(df_final, df_mes, on='Nome Poço ANP', how='outer')

    except FileNotFoundError:
        print(f"Arquivo {nome_arquivo_entrada} não encontrado. Pulando...")

# Exporta a única planilha unificada final
if df_final is not None:
    df_final.to_csv("ATUALIZADO_CONSOLIDADO_GERAL" + formato, index=False, sep=';', encoding='latin-1')
    print("Planilha unificada gerada com sucesso!")
else:
    print("Nenhum dado foi processado.")

display(df_final)
```
